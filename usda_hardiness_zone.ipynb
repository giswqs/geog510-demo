```python
from geopy.geocoders import Nominatim

geolocator = Nominatim(user_agent="geo_address_locator")
address = input("Enter an address: ")

try:
    # addressdetails=True is required to get the postal code reliably
    location = geolocator.geocode(address, addressdetails=True)

    if location:
        # 1. Get Latitude and Longitude
        lat = location.latitude
        lon = location.longitude

        # 2. Extract ZIP code from the 'raw' metadata
        address_data = location.raw.get("address", {})
        zip_code = address_data.get("postcode", "Not Found")

        # Formatting Output
        print("-" * 30)
        print(f"Latitude:  {lat}")
        print(f"Longitude: {lon}")
        print(f"ZIP Code:  {zip_code}")
        print("-" * 30)
        print(f"Full Result: {location.address}")

    else:
        print("Could not find that location.")

except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import folium


# Create a map centered at the coordinates
m = folium.Map(location=[location.latitude, location.longitude], zoom_start=15)
folium.Marker([location.latitude, location.longitude], popup="Your Address").add_to(m)

m  # This displays the map directly in your notebook
```

```python
import requests


def get_hardiness_zone(zip_code):
    url = f"https://phzmapi.org/{zip_code}.json"
    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        return data["zone"]
    return None
```

```python
# Example usage
print(f"The hardiness zone is: {get_hardiness_zone(zip_code)}")
```
